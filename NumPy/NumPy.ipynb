# Basic Program of NumPy


```python
import numpy as np 
X=np.array([[1.0,200.0],[2.0,300.0],[3.0,400.0]])
print(X.shape)
mu=X.mean(axis=0)
sd=X.std(axis=0)
Z=(X-mu)/sd
print(f"Mean {mu} Standard deviation {sd.round(3)}")
print(Z.round(3))
print(X @ np.array([0.5,0.01]))
```
